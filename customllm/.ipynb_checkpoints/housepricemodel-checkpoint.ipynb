```python
from sklearn import datasets
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
```

```python
data_house = datasets.fetch_california_housing()
```

```python
data_house_df = pd.DataFrame(data_house["data"],columns=data_house.feature_names)
data_house_df.head()
```

```python
data_house_df[:5]
```

```python
data_house
```

```python
data_house_df.shape
```

```python
data_house_df["target"] = pd.DataFrame(data_house["target"],columns=data_house.target_names)
```

```python
data_house_df.head()
```

```python
data_house_df.head()
```

```python
X = data_house_df.drop(columns=["target"])
X.head()
```

```python
y = data_house_df["target"]
y.head()
```

```python
y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
clf = DecisionTreeRegressor()

clf.fit(X_train,y_train)

score = clf.score(X_test,y_test)

print(f"Test Model Score :{score}")
```

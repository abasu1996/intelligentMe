```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier

heart_disease_data = pd.read_csv('scikit-learn-data/heart-disease.csv')
heart_disease_data
```

```python
# Create X feature
X = heart_disease_data.drop('target', axis=1)
# Create y feature
y = heart_disease_data['target']
X
```

```python
from sklearn.ensemble import RandomForestClassifier

clf = RandomForestClassifier()
clf.get_params()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2)
X_train
```

```python
X_test
```

```python
clf.fit(X_train, y_train)
```

```python
y_preds = clf.predict(X_test)
y_preds
```

```python
X_test
```

```python
clf.score(X_train, y_train)
```

```python
clf.score(X_test,y_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

print(classification_report(y_test, y_preds))
```

```python
confusion_matrix(y_test, y_preds)
```

```python
accuracy_score(y_test, y_preds)
```

```python
y_preds
```

```python
## Improve a model
##1. Try different amount of n_estimators
np.random.seed(42)
for i in range(10,100,10):
    print(f'Trying model with {i} estimators')
    clf = RandomForestClassifier(n_estimators=i).fit(X_train, y_train)
    print(f'Model accuracy on test set: {clf.score(X_test, y_test)*100:.2f}%')
    print('')
```

```python
import pickle

pickle.dump(clf,open('random_pickle_model.pkl','wb'))
```

```python
loaded_model = pickle.load(open('random_pickle_model.pkl','rb'))
loaded_model.score(X_test,y_test)
```

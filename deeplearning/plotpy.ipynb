### Make figures with Numpy figures

* Line Plot
* Scatter Plot
* Bar Plot
* Histogram
* Subplots

```python
## create some data
x = np.linspace(0,10,100)
x
```

```python
## Plot the data
fig,ax = plt.subplots()
ax.plot(x,x**2 );
```

```python
## use same data to create a scatter plot
fig,ax = plt.subplots()
ax.scatter(x,x**2 );
```

```python
nut_butter_prices = {
    "Almond Butter":10,
    "Peanut Butter":8,
    "Cashew Butter":12
}
fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(),nut_butter_prices.values());
ax.set(title="Nut Butter Prices",ylabel="Price ($)",xlabel="Type of Butter");
```

```python
fig,ax = plt.subplots()
ax.barh(nut_butter_prices.keys(),list(nut_butter_prices.values()));
ax.set(title="Nut Butter Prices",xlabel="Price ($)",ylabel="Type of Butter");
```

```python
fig,((ax1,ax2),(ax3,ax4)) = plt.subplots(nrows=2,ncols=2,figsize=(10,5))
ax1.plot(x,x/2);
ax1.set(title="Line Plot")
ax2.scatter(nut_butter_prices.keys(),nut_butter_prices.values());
ax2.set(title="Scatter Plot")
ax3.bar(nut_butter_prices.keys(),nut_butter_prices.values());
ax3.set(title="Bar Plot")
ax4.barh(nut_butter_prices.keys(),list(nut_butter_prices.values()));
ax4.set(title="Horizontal Bar Plot")
```

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 5))

ax[0, 0].barh(list(nut_butter_prices.keys()), nut_butter_prices.values());
ax[0, 0].set(title="Bar Plot")
```
